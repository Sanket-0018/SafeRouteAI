```python
import pandas as pd
import numpy as np

DATA_PATH = "../data/raw/indian_roads_dataset.csv"

df = pd.read_csv(DATA_PATH)

print("Dataset loaded successfully!")
print("Shape:", df.shape)

df.head()
```

```python
print("========== DATASET OVERVIEW ==========")

print("\nShape:")
print(df.shape)

print("\nColumn Names:")
for i, col in enumerate(df.columns, start=1):
    print(f"{i}. {col}")

print("\nData Types:")
print(df.dtypes)

print("\nMissing Values:")
print(df.isnull().sum())

print("\nDuplicate Rows:")
print(df.duplicated().sum())

print("\nUnique Values:")
print(df.nunique())

print("\nDataset Info:")
df.info()
```

```python
print("========== NUMERICAL SUMMARY ==========")
display(df.describe(include="all").T)
```

```python
categorical_columns = [
    "city",
    "state",
    "day_of_week",
    "road_type",
    "traffic_signal",
    "weather",
    "visibility",
    "traffic_density",
    "cause",
    "accident_severity",
    "is_weekend",
    "is_peak_hour",
    "festival"
]

for col in categorical_columns:
    print(f"\n========== {col.upper()} ==========")
    print(df[col].value_counts(dropna=False))
```

```python
print("========== CITY DISTRIBUTION ==========")

city_counts = df["city"].value_counts()

print(city_counts)

print("\nPercentage:")
print((city_counts / len(df) * 100).round(2))
```

```python
print("========== CITY-STATE RELATIONSHIP ==========")

city_state = pd.crosstab(df["city"], df["state"])

display(city_state)
```

```python
print("========== GEOGRAPHIC DISTRIBUTION BY CITY ==========")

geo_summary = df.groupby("city").agg(
    records=("accident_id", "count"),
    min_latitude=("latitude", "min"),
    max_latitude=("latitude", "max"),
    min_longitude=("longitude", "min"),
    max_longitude=("longitude", "max"),
    avg_latitude=("latitude", "mean"),
    avg_longitude=("longitude", "mean")
)

display(geo_summary)
```

```python
print("========== RISK SCORE VALUES ==========")

risk_values = df["risk_score"].value_counts().sort_index()

display(risk_values)

print("\nNumber of unique risk scores:", df["risk_score"].nunique())
```

```python
print("========== RISK SCORE BY SEVERITY ==========")

display(
    df.groupby("accident_severity")["risk_score"]
      .agg(["count", "min", "mean", "median", "max"])
)
```

```python
print("========== RISK SCORE RELATIONSHIPS ==========")

risk_relationships = df.groupby(
    ["accident_severity", "traffic_density"]
)["risk_score"].agg(["count", "mean", "min", "max"])

display(risk_relationships)
```

```python
print("========== RISK SCORE VS IMPORTANT VARIABLES ==========")

numeric_cols = [
    "hour",
    "lanes",
    "traffic_signal",
    "temperature",
    "vehicles_involved",
    "casualties",
    "is_peak_hour",
    "risk_score"
]

display(df[numeric_cols].corr(numeric_only=True)["risk_score"].sort_values(ascending=False))
```

```python
print("========== RISK SCORE BY CITY ==========")

display(
    df.groupby("city")["risk_score"]
      .agg(["count", "mean", "min", "max"])
      .sort_values("mean", ascending=False)
)
```

```python
print("========== DUPLICATE FEATURE COMBINATIONS ==========")

feature_cols = [
    "city",
    "road_type",
    "lanes",
    "traffic_signal",
    "weather",
    "visibility",
    "traffic_density",
    "cause",
    "accident_severity",
    "vehicles_involved",
    "casualties",
    "is_peak_hour"
]

risk_consistency = (
    df.groupby(feature_cols)["risk_score"]
      .nunique()
)

print("Feature combinations with multiple risk scores:",
      (risk_consistency > 1).sum())

print("Total unique feature combinations:",
      len(risk_consistency))
```

```python
print("========== COORDINATE ANALYSIS ==========")

print("Unique latitude:", df["latitude"].nunique())
print("Unique longitude:", df["longitude"].nunique())

print(
    "Exact latitude-longitude pairs:",
    df[["latitude", "longitude"]].drop_duplicates().shape[0]
)
```

```python
print("========== DATE RANGE ==========")

date_series = pd.to_datetime(df["date"], errors="coerce")

print("Minimum date:", date_series.min())
print("Maximum date:", date_series.max())
print("Invalid dates:", date_series.isna().sum())
```

```python
print("========== RISK SCORE BY CATEGORICAL FEATURES ==========")

categorical_features = [
    "road_type",
    "weather",
    "visibility",
    "traffic_density",
    "cause",
    "accident_severity",
    "is_peak_hour",
    "is_weekend"
]

for col in categorical_features:
    print(f"\n--- {col} ---")
    result = (
        df.groupby(col)["risk_score"]
        .agg(["count", "mean", "median", "min", "max"])
        .sort_values("mean", ascending=False)
    )
    display(result)
```

```python
print("========== RISK SCORE BY SEVERITY + TRAFFIC + PEAK HOUR ==========")

result = (
    df.groupby(
        ["accident_severity", "traffic_density", "is_peak_hour"]
    )["risk_score"]
    .agg(["count", "mean", "median", "min", "max"])
    .reset_index()
)

display(result)
```

```python
print("========== RISK SCORE DISTRIBUTION ==========")

display(
    df["risk_score"]
    .value_counts()
    .sort_index()
    .to_frame("count")
)

print("\nMean:", df["risk_score"].mean())
print("Median:", df["risk_score"].median())
print("Standard deviation:", df["risk_score"].std())
```

```python
risk_values = sorted(df["risk_score"].unique())

print("Risk score values:")
print(risk_values)

print("\nDifferences between consecutive values:")
print(np.diff(risk_values))
```

```python
print("========== WEATHER VS VISIBILITY ==========")

weather_visibility = pd.crosstab(
    df["weather"],
    df["visibility"],
    normalize="index"
) * 100

display(weather_visibility.round(2))
```

```python
print("========== TIME FIELD CONSISTENCY ==========")

date_check = pd.to_datetime(df["date"], errors="coerce")

calculated_day = date_check.dt.day_name()
calculated_weekend = calculated_day.isin(["Saturday", "Sunday"]).astype(int)

print("Day-of-week mismatches:",
      (calculated_day != df["day_of_week"]).sum())

print("Weekend mismatches:",
      (calculated_weekend != df["is_weekend"]).sum())

print("\nHour/time mismatches:")

calculated_hour = pd.to_datetime(
    df["time"],
    format="%H:%M",
    errors="coerce"
).dt.hour

print((calculated_hour != df["hour"]).sum())
```

```python
print("========== PEAK HOUR PATTERN ==========")

display(
    pd.crosstab(
        df["hour"],
        df["is_peak_hour"],
        normalize="index"
    ).round(2)
)
```

```python
print("========== CITY COORDINATE RANGES ==========")

geo_summary = df.groupby("city").agg(
    records=("accident_id", "count"),
    min_lat=("latitude", "min"),
    max_lat=("latitude", "max"),
    min_lon=("longitude", "min"),
    max_lon=("longitude", "max"),
    mean_lat=("latitude", "mean"),
    mean_lon=("longitude", "mean")
)

display(geo_summary)
```

```python
print("========== RISK SCORE VALUES ==========")

risk_values = sorted(df["risk_score"].unique())

print("Number of unique values:", len(risk_values))
print("Values:", risk_values)

print("\nDifferences:")
print(np.diff(risk_values))
```

```python
print("========== RISK SCORE BY CORE CONDITIONS ==========")

core_analysis = (
    df.groupby(
        [
            "weather",
            "traffic_density",
            "is_peak_hour",
            "accident_severity"
        ]
    )["risk_score"]
    .agg(["count", "mean", "min", "max"])
    .reset_index()
)

display(core_analysis)
```
